```python
import pandas as pd
import numpy as np
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt

df=pd.read_excel('data_for_project_1.xlsx',sheet_name='2020')

Y=np.array(df['life_expectancy'])
X_data=df[df.columns[1:]].values
```

```python
df
```

```python
def regression(df,var):
    Y=np.array(df['life_expectancy'])
    X_data=df[var]
    a=np.array([[1] for i in Y])
    X=np.concatenate((a,X_data),axis=1)
    coefs=np.linalg.pinv(X)@Y
    Y_pred=X@coefs
    r2=r2_score(Y, Y_pred)
    print(f'coefs={coefs}')
    print(f'R^2={r2}')
    return coefs,X,Y

def line(coefs,X):
  x=np.linspace(min(X[:,1]),max(X[:,1]),2)
  y=coefs[0]+x*coefs[1]
  return x,y
```

Модель, описывающая взаимосвязь между средней продолжительностью жизни и загрязнением окружающего мира, уровнем заболеваимости, численностью населения на одну больничную койку,средним доходом и количеством докторов.

```python
coefs1,X1,Y1=regression(df,['pollution','disease_rate','hosp_bed','av_income','doctors'])
```

Модель, описывающая взаимосвязь между средней продолжительностью жизни и загрязнением окружающего мира.

```python
coefs2,X2,Y2=regression(df,['pollution'])
x2,y2=line(coefs2,X2)

fig2,ax2=plt.subplots()
ax2.set_facecolor('black')
ax2.scatter(df['pollution'].values,df['life_expectancy'].values)
ax2.plot(x2,y2,c='orange')
ax2.set_xlabel('pollution')
ax2.set_ylabel('life expectancy')
```
